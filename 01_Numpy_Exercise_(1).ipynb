# NumPy Exercises

Now that we've learned about NumPy let's test your knowledge. We'll start off with a few simple tasks and then you'll be asked some more complicated questions.

#### Import NumPy as np

```python
import numpy as np

```

#### Create an array of 10 zeros

```python
zero_array = np.zeros(10)
print (zero_array)
```

#### Create an array of 10 ones

```python
ones_array = np.ones(10)
print (ones_array)
```

#### Create an array of 10 fives

```python
fives_array = np.ones(10)*5
print (fives_array)
```

#### Create an array of the integers from 10 to 50

```python
range_array = np.arange(10,51)
print (range_array)
```

#### Create an array of all the even integers from 10 to 50

```python
even_array = np.arange(10,51,2)
print (even_array)
```

#### Create a 3x3 matrix with values ranging from 0 to 8

```python
range_array = np.arange(0,9)
range_array.reshape(3,3)
```

#### Create a 3x3 identity matrix

```python
np.matrix_transpose(np.eye(3))
```

#### Use NumPy to generate a random number between 0 and 1

```python
random_number = np.random.rand(1)
print (random_number)
```

#### Use NumPy to generate an array of 25 random numbers sampled from a standard normal distribution

```python
random_number = np.random.randn(25)
print (random_number)
```

#### Create the following matrix:

```python
matrix = np.arange(0.01,1.01,0.01)
matrix.reshape(10,10)
```

#### Create an array of 20 linearly spaced points between 0 and 1:

```python
linearly_spaced = np.linspace(0,1,20)
print (linearly_spaced)
```

## Numpy Indexing and Selection

Now you will be given a few matrices, and be asked to replicate the resulting matrix outputs:

```python
mat = np.arange(1,26).reshape(5,5)
mat
```

### Now do the following

#### Get the sum of all the values in mat

```python
sum_of_valus = np.sum(mat)
print (sum_of_valus)
```

#### Get the standard deviation of the values in mat

```python
from typing import ValuesView
ValuesView = np.std(mat)
print (ValuesView)
```

#### Get the sum of all the columns in mat

```python
sum_of_columns = np.sum(mat, axis=0)
print (sum_of_columns)
```
